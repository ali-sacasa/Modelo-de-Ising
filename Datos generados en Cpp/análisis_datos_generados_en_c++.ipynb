```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

import numpy as np
import matplotlib.pyplot as plt


# ================================================================
# (1) DEFINICIÓN ROBUSTA Y FIJA DEL VECTOR DE TIEMPOS
# ================================================================

T = np.array([
0, 0.05, 0.1, 0.15, 0.2, 0.25, 0.3, 0.35, 0.4, 0.45,
0.5, 0.55, 0.6, 0.65, 0.7, 0.75, 0.8, 0.85, 0.9, 0.95,
1, 1.05, 1.1, 1.15, 1.2, 1.25, 1.3, 1.35, 1.4, 1.45,
1.5, 1.55, 1.6, 1.65, 1.7, 1.75, 1.8, 1.85, 1.9, 1.95,
2, 2.05, 2.1, 2.15, 2.2, 2.25, 2.3, 2.35, 2.4, 2.45,
2.5, 2.55, 2.6, 2.65, 2.7, 2.75, 2.8, 2.85, 2.9, 2.95,
3, 3.05, 3.1, 3.15, 3.2, 3.25, 3.3, 3.35, 3.4, 3.45,
3.5, 3.55, 3.6, 3.65, 3.7, 3.75, 3.8, 3.85, 3.9, 3.95,
4
])


# ================================================================
# (2) FUNCIONES DEL MODELO DE ISING
# ================================================================

def magnetization_ising1d_exact(beta, N=100, J=1.0):
    """
    Solución exacta del Ising 1D sin campo.
    m = 0 para campo h = 0 (simetría Z_2)
    Energía libre y correlaciones sí dependen de beta y J.
    """
    lam_plus  = np.cosh(beta*J) + np.sqrt(np.sinh(beta*J)**2 + 1)
    lam_minus = np.cosh(beta*J) - np.sqrt(np.sinh(beta*J)**2 + 1)
    Z = (lam_plus**N + lam_minus**N)
    f = -(1/beta) * np.log(lam_plus)  # energía libre por spin
    return 0*T + 0, f  # magnetización=0, energía libre=f constante


def magnetization_ising2d_exact(beta, J=1.0):
    """
    Fórmula exacta de Onsager para la magnetización espontánea en 2D.
    Para T > Tc, m = 0.
    Para T < Tc:    m = [1 - sinh^{-4}(2βJ)]^{1/8}
    """
    Tc = 2*J / np.log(1 + np.sqrt(2))
    T = 1/beta

    m = np.zeros_like(T)
    below = T < Tc

    sinh_arg = np.sinh(2 * beta[below] * J)
    m[below] = (1 - 1/(sinh_arg**4))**(1/8)
    return m, Tc



# ================================================================
# (3) GENERACIÓN DE DATOS ANALÍTICOS PARA T DE 0 A 4
# ================================================================

beta = 1/(T + 1e-9 + 1)     # Para evitar T=0 → solo una convención físico-numérica
J = 1.0

m1d, f1d = magnetization_ising1d_exact(beta, N=200, J=J)
m2d, Tc  = magnetization_ising2d_exact(beta, J=J)


# ================================================================
# (4) GRÁFICAS
# ================================================================

plt.figure(figsize=(10,6))
plt.plot(T, m1d, '--', label="Ising 1D (exacto)")
plt.plot(T, m2d, '-', label="Ising 2D (Onsager exacto)")
plt.axvline(Tc, color="gray", linestyle="--", label=f"Tc (2D) = {Tc:.3f}")
plt.xlabel("T")
plt.ylabel("Magnetización")
plt.title("Comparación Ising 1D vs 2D — Soluciones exactas")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()


plt.figure(figsize=(10,6))
plt.plot(T, f1d, label="Energía libre por spin (1D)")
plt.xlabel("T")
plt.ylabel("f(T)")
plt.title("Energía libre exacta del Ising 1D")
plt.grid(True)
plt.tight_layout()
plt.show()



# ================================================================
# (5) INFO FINAL
# ================================================================
print("\nCálculo terminado.")
print(f"Temperatura crítica exacta 2D: Tc = {Tc:.6f}")
print("Gráficas generadas correctamente.\n")
```
