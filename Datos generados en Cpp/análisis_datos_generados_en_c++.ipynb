```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import csv
import os

plt.style.use("seaborn-v0_8-darkgrid")


# ================================================================
# VECTOR DE TIEMPO DEFINIDO MANUALMENTE (evita parsing del CSV)
# ================================================================
T = np.array([
0, 0.05, 0.1, 0.15, 0.2, 0.25, 0.3, 0.35, 0.4, 0.45,
0.5, 0.55, 0.6, 0.65, 0.7, 0.75, 0.8, 0.85, 0.9, 0.95,
1, 1.05, 1.1, 1.15, 1.2, 1.25, 1.3, 1.35, 1.4, 1.45,
1.5, 1.55, 1.6, 1.65, 1.7, 1.75, 1.8, 1.85, 1.9, 1.95,
2, 2.05, 2.1, 2.15, 2.2, 2.25, 2.3, 2.35, 2.4, 2.45,
2.5, 2.55, 2.6, 2.65, 2.7, 2.75, 2.8, 2.85, 2.9, 2.95,
3, 3.05, 3.1, 3.15, 3.2, 3.25, 3.3, 3.35, 3.4, 3.45,
3.5, 3.55, 3.6, 3.65, 3.7, 3.75, 3.8, 3.85, 3.9, 3.95,
4
])


# ================================================================
# UTILITY: Robust CSV loader (fix for malformed C++ files)
# ================================================================
def robust_read_csv(path):
    if not os.path.exists(path):
        print(f"[WARN] File not found: {path}")
        return None

    try:
        return pd.read_csv(path)
    except:
        pass

    try:
        with open(path, "r") as f:
            sample = f.read(2048)
            dialect = csv.Sniffer().sniff(sample, delimiters=",;:\t ")
        return pd.read_csv(path, delimiter=dialect.delimiter)
    except:
        pass

    try:
        data = {}
        with open(path, "r") as f:
            for line in f:
                if "=" in line:
                    k, v = line.strip().split("=")
                    data[k] = v
        if len(data) > 0:
            return pd.DataFrame([data])
    except:
        pass

    print(f"[ERROR] Could not parse {path}")
    return None


# ================================================================
# LOAD ALL CSV FILES
# ================================================================
config = robust_read_csv("config.csv")
time_raw = robust_read_csv("time.csv")  # NOT USED for T, only for debugging

energy          = robust_read_csv("energy_results.csv")
magnetization   = robust_read_csv("magnetization.csv")
fidelity        = robust_read_csv("fidelity_evolution.csv")
metrics         = robust_read_csv("metrics.csv")
variance        = robust_read_csv("variance.csv")
local_energy_full = robust_read_csv("local_energy_full.csv")
local_energy_proj = robust_read_csv("local_energy_proj.csv")
ricci_beta      = robust_read_csv("ricci_beta_list.csv")
ricci_fid       = robust_read_csv("ricci_fidelitites.csv")
states_full     = robust_read_csv("states_full_amplitudes.csv")
states_proj     = robust_read_csv("states_proj_amplitudes.csv")

print(energy)
# ================================================================
# FIGURE 1 — ENERGY EVOLUTION
# ================================================================
plt.figure(figsize=(10, 6))

plt.plot(T, energy["energy_proj"], label="Proyectado", linestyle="--", linewidth=2)
plt.xlabel("Tiempo")
plt.ylabel("Energía total")
plt.title("Evolución de Energía")
plt.legend()
plt.grid(True, alpha=0.3)
plt.savefig("plot_energy_total.png", dpi=150)
plt.close()


# ================================================================
# FIGURE 2 — ENERGY COMPONENTS + MAGNETIZATION + FIDELITY
# ================================================================
plt.figure(figsize=(12, 8))

# Full components
plt.subplot(2, 2, 1)
plt.plot(T, energy["E_ZZ_full"], label="ZZ", linewidth=2)
plt.plot(T, energy["E_X_full"], label="X", linewidth=2)
plt.title("Componentes energía (Completo)")
plt.xlabel("Tiempo")
plt.ylabel("E")
plt.legend()

# Projected components
plt.subplot(2, 2, 2)
plt.plot(T, energy["E_ZZ_proj"], label="ZZ proj", linewidth=2)
plt.plot(T, energy["E_X_proj"], label="X proj", linewidth=2)
plt.title("Componentes energía (Proyectado)")
plt.xlabel("Tiempo")
plt.ylabel("E")
plt.legend()

# Magnetization
plt.subplot(2, 2, 3)
plt.plot(T, magnetization["magnetization_full"], label="Completo", linewidth=2)
plt.plot(T, magnetization["magnetization_proj"], "--", label="Proyectado", linewidth=2)
plt.xlabel("Tiempo")
plt.ylabel("⟨Sz⟩ por sitio")
plt.title("Magnetización")
plt.legend()

# Fidelity
plt.subplot(2, 2, 4)
plt.plot(T, fidelity["fidelity"], "g-", linewidth=2)
plt.xlabel("Tiempo")
plt.ylabel("Fidelidad")
plt.title("Fidelidad")
plt.ylim(0, 1.05)

plt.tight_layout()
plt.savefig("plot_summary_energy_mag_fid.png", dpi=150)
plt.close()


# ================================================================
# FIGURE 3 — LOCAL ENERGY PROFILES
# ================================================================
plt.figure(figsize=(12, 6))
for col in local_energy_full.columns[1:]:
    plt.plot(T, local_energy_full[col], alpha=0.7)
plt.title("Energía local por sitio (Completo)")
plt.xlabel("Tiempo")
plt.ylabel("E_local")
plt.grid(True)
plt.savefig("plot_local_energy_full.png", dpi=150)
plt.close()

plt.figure(figsize=(12, 6))
for col in local_energy_proj.columns[1:]:
    plt.plot(T, local_energy_proj[col], alpha=0.7)
plt.title("Energía local por sitio (Proyectado)")
plt.xlabel("Tiempo")
plt.ylabel("E_local")
plt.grid(True)
plt.savefig("plot_local_energy_proj.png", dpi=150)
plt.close()


# ================================================================
# FIGURE 4 — VARIANCE AND METRICS
# ================================================================
plt.figure(figsize=(12, 5))
plt.plot(T, variance["variance_full"], label="Var Full", linewidth=2)
plt.plot(T, variance["variance_proj"], label="Var Proj", linewidth=2)
plt.title("Varianza de la energía")
plt.xlabel("Tiempo")
plt.ylabel("Varianza")
plt.legend()
plt.savefig("plot_variance.png", dpi=150)
plt.close()

plt.figure(figsize=(12, 5))
for col in metrics.columns[1:]:
    plt.plot(T, metrics[col], label=col)
plt.title("Métricas numéricas (Norma, Overlap, etc.)")
plt.xlabel("Tiempo")
plt.legend()
plt.savefig("plot_metrics.png", dpi=150)
plt.close()


# ================================================================
# FIGURE 5 — RICCI FLOW
# ================================================================
plt.figure(figsize=(12, 5))
plt.plot(T, ricci_beta["beta"], linewidth=2)
plt.title("Beta del flujo de Ricci")
plt.xlabel("Tiempo")
plt.ylabel("β(t)")
plt.savefig("plot_ricci_beta.png", dpi=150)
plt.close()

plt.figure(figsize=(12, 5))
plt.plot(T, ricci_fid["fidelity"], linewidth=2)
plt.title("Fidelidad inducida por Ricci Flow")
plt.xlabel("Tiempo")
plt.ylabel("Fidelidad")
plt.ylim(0, 1.05)
plt.savefig("plot_ricci_fidelity.png", dpi=150)
plt.close()


# ================================================================
# FIGURE 6 — AMPLITUDES DEL ESTADO
# ================================================================
def plot_state_amplitudes(df, title, fname):
    plt.figure(figsize=(10, 6))
    for col in df.columns[1:]:
        plt.plot(T, df[col], alpha=0.5)
    plt.title(title)
    plt.xlabel("Tiempo")
    plt.ylabel("Re(ψ_i) o |ψ_i|²")
    plt.savefig(fname, dpi=150)
    plt.close()

plot_state_amplitudes(states_full, "Amplitudes del estado (Completo)", "plot_states_full.png")
plot_state_amplitudes(states_proj, "Amplitudes del estado (Proyectado)", "plot_states_proj.png")


print("\n✔ Todas las gráficas fueron generadas correctamente.\n")
```
